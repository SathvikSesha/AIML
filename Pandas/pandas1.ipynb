```python
import pandas as pd

df = pd.DataFrame({
    "Name":["A","B","C","D","E"],
    "Age":[18,20,19,21,22],
    "Marks":[95,70,85,60,40],
    "City":["Guntur","Hyderabad","Guntur","Vijayawada","Hyderabad"]
})

print(df.shape)
print(df.columns)
print(df.iloc[0:3,])
print(df.iloc[-1:-3:-1,])
print(df[['Name','Marks']])
print('Minimum marks:',df['Marks'].min())
print('Maximum marks:',df['Marks'].max())
print('Average:',df['Marks'].mean())
print(df[df['Marks']>80]['Name'])
print(df[df['Age']<20][['Name','Age']])
df.sort_values('Marks',ascending = False,inplace = True)
print(df)
df['pass'] = ["P" if x else "F" for x in df['Marks']>50]
print(df)
print(df.loc[2])
print(df.loc[3])
print(df.iloc[0:3,])
x = df['Marks'].argmax();
print(df.iloc[x])
print(df.at[2,'Name'])
```

```python
df = pd.DataFrame({
    "Name":["A","B","C","D","E","F"],
    "City":["Guntur","Hyderabad","Guntur","Hyderabad","Guntur","Vijayawada"],
    "Marks":[95,70,85,60,90,80],
    "Gender":["M","F","M","F","M","F"]
})
```

```python
print(df.groupby('City')['Marks'].mean())
print(df.groupby('City')['Marks'].max())
print(df.groupby('City')['Marks'].sum())
print(df.groupby('City')['Name'].count())
print(df.groupby('City')['Marks'].agg(['mean','max','min','count']))
```

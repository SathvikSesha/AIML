```python
import requests as req
```

```python
URL = "https://www.scrapethissite.com/pages/simple/"
```

```python
res = req.get(URL)
```

```python
print(res.status_code)

if res.status_code == 200:
    print(res.headers)

    with open("Scrap_data/data1.html", "w", encoding="utf-8") as f:
        f.write(res.text)
```

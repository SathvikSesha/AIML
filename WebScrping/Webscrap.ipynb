```python
from bs4 import BeautifulSoup
with open("Scrap_data/data1.html","r") as f:
    html_data = f.read()
soup = BeautifulSoup(html_data,"lxml")
```

```python
# there are several methods that can be used in the extraction of data using beautiful soup
soup.find("h1")
countries = soup.find_all("h3")
names = []
population = []
for country in countries:
    name = country.get_text(strip = True)
    # print(country.find_parent("div"))
    p = country.find_next("div").select("span.country-population")[0].get_text(strip = True)
    names.append(name)
    population.append(p)
info = {"Names":names,"Population":population}
# print(info)
import pandas as pd
df = pd.DataFrame(info)
df["Population"] = pd.to_numeric(df["Population"])
print(df)
print(df[df['Names'].str.lower()=="india"])
print(df.sort_values('Population',ascending = False))
```

```python
df.to_csv("Scrap_data/cleaned_data.csv",index = False)
```

```python
import requests as req
from bs4 import BeautifulSoup

i = 1
authors = []
quotes = []

while True:
    data = req.get(f"https://quotes.toscrape.com/page/{i}/")
    soup = BeautifulSoup(data.text, 'lxml')

    quote_blocks = soup.select("div.quote")

    if not quote_blocks:
        break

    for quote in quote_blocks:
        tags = [t.get_text(strip=True) for t in quote.select("a.tag")]

        if "life" in tags:
            quotes.append(
                quote.select_one("span.text").get_text(strip=True)
            )

            authors.append(
                quote.select_one("small.author").get_text(strip=True)
            )

    i += 1

import pandas as pd
df = pd.DataFrame({"Author":authors,"Quote":quotes})
print(df)
```
